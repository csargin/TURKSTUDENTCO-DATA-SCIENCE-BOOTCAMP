```python
"""
Bu projede, nesne yönelimli programlama (OOP) tekniklerini kullanarak bir
"Çevrimiçi Market Alışveriş Sepeti" oluşturacaksınız. Bu program, product.txt
dosyasını veri tabanı olarak kullanacak ve her satır bir ürünü temsil edecektir.
Her satırda ürün adı, kategori, fiyat ve stok miktarı virgülle ayrılmış şekilde
tutulacaktır. Proje, öğrendiklerinizi pratikte uygulayarak pekiştirmenize
yardımcı olacaktır.

1. Market Sınıfı Oluşturma
  a. Sınıf Tanımı ve Yapıcı Metot (Constructor):
    • Market adında bir sınıf oluşturun.
    • Yapıcı (constructor) metot, product.txt dosyasını açmalı ve dosyanın
      mevcut olup olmadığını kontrol etmelidir. Dosya yoksa oluşturulmalıdır.
  b. Yıkıcı Metot (Destructor):
    • Yıkıcı metot, program sonlandığında product.txt dosyasını kapatmalıdır.
2. Market Sınıfına Eklenecek Metotlar
  a. List Product: Bu metot, product.txt dosyasındaki tüm ürünleri
    listeleyecektir.
    • Dosya içeriğini okuyun.
    • splitlines() metodu ile her satırı bir listeye ekleyin.
    • Listenin her elemanı bir ürünü temsil edecektir. Bu bilgileri kullanarak
      ürün adlarını, kategorilerini, fiyatlarını ve stok miktarlarını yazdırın.
  b. Add Product: Bu metot, product.txt dosyasına yeni bir ürün ekleyecektir.
    • Kullanıcıdan ürün adı, kategori, fiyat ve stok miktarını girmesini isteyin.
    • Bu bilgilerle bir satır oluşturun ve virgülle ayrılmış şekilde düzenleyin.
    • Bu satırı dosyaya ekleyin.
  c. Delete Product: Bu metot, eklenen ürünü product.txt dosyasından silecektir.
    • Kullanıcıdan ürün adını veya numarasını girmesini isteyin.
    • Dosya içeriğini okuyup her satırı bir listeye ekleyin.
    • Silinecek ürünün listede hangi satırda olduğunu bulun.
    • Ürünü listeden çıkarın.
    • Dosya içeriğini temizleyip, güncellenmiş listeyi yeniden dosyaya yazın.
3. Menu: Kullanıcının etkileşim kurabileceği bir menu oluşturun.
  • Market sınıfından market adında bir nesne oluşturun.
  • market nesnesini arayüz menüsüne bağlayın.
  • Kullanıcıdan menu öğesi için bir giriş isteyin ve bu girdiyi bir değişkene
    atayın.
  • if-elif-else yapısını kullanarak kullanıcı girişini kontrol edin ve ilgili
    metotları çalıştırın.

*** MENÜ ***
1) Ürünleri Listele
2) Ürün Ekle
3) Ürün Sil
4) Çıkış

"""

class Market:
    def __init__(self, filename="product.txt"):
        self.filename = filename
        try:
            self.file = open(self.filename, "r+")
        except FileNotFoundError:
            self.file = open(self.filename, "w+")
            print(f"{self.filename} dosyası oluşturuldu.")

    def __del__(self):
        if self.file:
            self.file.close()
            print(f"{self.filename} dosyası kapatıldı.")

    def list_products(self):
        self.file.seek(0)  # Dosya başlangıcına dön
        products = self.file.readlines()
        if not products:
            print("Ürün bulunamadı.")
            return

        print("Ürün Listesi:")
        for i, product in enumerate(products):
            try:
                name, category, price, stock = product.strip().split(",")
                print(f"{i+1}. Ürün Adı: {name}, Kategori: {category}, Fiyat: {price}, Stok: {stock}")
            except ValueError:
                print(f"Hatalı ürün formatı: {product.strip()}")

    def add_product(self):
        name = input("Ürün adı: ")
        category = input("Kategori: ")
        price = input("Fiyat: ")
        stock = input("Stok miktarı: ")
        product_line = f"{name},{category},{price},{stock}\n"
        self.file.write(product_line)
        print("Ürün eklendi.")

    def delete_product(self):
        self.file.seek(0)
        products = self.file.readlines()
        self.list_products() # Ürünleri listele, kullanıcı silmek istediği ürünü görsün
        try:
          product_index = int(input("Silmek istediğiniz ürünün numarasını girin: ")) - 1
          if 0 <= product_index < len(products):
              del products[product_index]
              self.file.seek(0)
              self.file.truncate(0) # Dosya içeriğini temizle
              self.file.writelines(products)
              print("Ürün silindi.")
          else:
              print("Geçersiz ürün numarası.")
        except (ValueError, IndexError):
            print("Geçersiz giriş.")

if __name__ == "__main__":
    market = Market()
    while True:
        print("\n*** MENÜ ***")
        print("1. Ürünleri listele")
        print("2. Ürün ekle")
        print("3. Ürün sil")
        print("4. Çıkış")

        choice = input("Seçiminizi girin (1-4): ")

        if choice == "1":
            market.list_products()
        elif choice == "2":
            market.add_product()
        elif choice == "3":
            market.delete_product()
        elif choice == "4":
            break
        else:
            print("Geçersiz seçim.")

    del market
```
